#### At an eCommerce company a parcel needs to be packed. It should contain the following 4 items stored in warehouses at different geo locations:

items = [
        {"id": 123, "wh": "erfurt", "geolocation": (50.9787, 11.03283)},
        {"id": 456, "wh": "stradella", "geolocation": (45.07445, 9.30169)},
        {"id": 789, "wh": "lahr", "geolocation": (48.34042, 7.86886)},
        {"id": 102, "wh": "gryfino", "geolocation": (53.25243, 14.48831)}]
        
#### Write a function that finds the warehouse where all the items should be collected, so that they have to be transported the smallest distance possible.

```python
import math
import sys
def find_minicost_collection(items:list)->int:
    cost = {}
    for itm in items:
        cost[itm["id"]] = 0
        for dest in items:
            cost[itm["id"]] += math.sqrt((itm["geolocation"][0] - dest["geolocation"][0])**2 +
                                        (itm["geolocation"][1] - dest["geolocation"][1])**2)

    mincost = sys.maxsize
    mindest = items[0]["id"]
    for key, itm in cost.items():
        if mincost < itm :
            mincost = itm
            mindest = key

    return mindest
```

```python
items = [
        {"id": 123, "wh": "erfurt", "geolocation": (50.9787, 11.03283)},
        {"id": 456, "wh": "stradella", "geolocation": (45.07445, 9.30169)},
        {"id": 789, "wh": "lahr", "geolocation": (48.34042, 7.86886)},
        {"id": 102, "wh": "gryfino", "geolocation": (53.25243, 14.48831)}
]
find_minicost_collection(items)
```

```python
def find_minicost_collection_faster(items:list)->int:
    cost = {}
    xsum = 0
    ysum = 0
    count = 0
    for itm in items:
        xsum += itm["geolocation"][0]
        ysum += itm["geolocation"][1]
        count += 1
    xmean = xsum/count
    ymean = ysum/count

    mincost = sys.maxsize
    mindest = items[0]["id"]

    for itm in items:
        cost = math.sqrt((itm["geolocation"][0] - xmean)**2 +
                         (itm["geolocation"][1] - ymean)**2)
        if cost < mincost:
            mincost = cost
            mindest = itm["id"]
    return mindest
```
