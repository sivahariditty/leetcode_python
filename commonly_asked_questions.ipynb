```python
def is_palidrome(s:str)->bool:
    n = len(s)
    j = n-1
    for i in range(n//2):
        if(s[i] != s[j]):
            return False
        j = j - 1
    return True
```

```python
#anagram
def is_anagram(s1:str, s2:str)->bool:
    for c in s1:
        if c not in s2:
            return False
    return True
# takes O(n^2)
```

```python
is_anagram('abcd','cbad')
```

```python
is_anagram('abcd','cbrd')
```

```python
def is_anagram_fast(s1:str, s2:str)->bool:
    s1 = sorted(s1)
    s2 = sorted(s2)
    if s1 == s2:
        return True
    else:
        return False
```

```python
is_anagram_fast('abcd','cbrd')
```

```python
is_anagram('abcd','cbad')
```

```python
#can we do it in O(n) with more space

def is_anagram_superfast(s1:str, s2:str)->bool:
    s1 = list(s1)
    s2 = list(s2)
    count = {}
    for c in s1:
        if c not in count:
            count[c] = 0
        count[c] += 1

    for d in s2:
        if d not in count:
            return False #not anagram
        else:
            count[d] = count[d] - 1

    for key,val in count:
        if val != 0:
            return False

    return True
```

```python
is_anagram('abcd','cbad')
```

```python
def subarray_sum(arr:list)->int:
    maxsum = arr[0]
    csum = 0
    for i in range(len(arr)):
        maxsum = arr[0]
        for j in range(i,len(arr)):
            csum = csum + arr[j]
            if csum > maxsum :
                maxsum = csum
    return maxsum
```

```python
subarray_sum([-2,-3,4,-1,-2,1,5,-3])
```

```python
def subarray_sum_fast(arr:list)->int:
    maxglobal, maxcurr = arr[0], arr[0]
    for i in range(1,len(arr)):
        maxcurr = max(maxcurr + arr[i], arr[i])
        if(maxcurr > maxglobal):
            maxglobal = maxcurr
    return maxglobal
#Kadane’s Algorithm runs in O(n)
```

```python
subarray_sum_fast([25,-3,-4,-1,-2,1,-5,-3])
```

```python
subarray_sum_fast([-2,-3,4,-1,-2,1,5,-3])
```

```python
#Move Zeros: Given an array, move all the zeros to the end while maintaining the relative order of the other elements.
[1,0,5,6,0,0,3,4,0]

def move_zeros(arr:list)->list:
    n = len(arr)
    ret_arr = []
    n_cnt = 0
    for i in range(n):
        if arr[i] != 0:
            ret_arr.append(arr[i])
            n_cnt += 1
    for j in range(n_cnt,n):
        ret_arr.append(0)
    return ret_arr
#takes O(n) time and O(n) space
```

```python
move_zeros([1,0,5,6,0,0,3,4,0])
```

```python
def move_zeros_fast(arr:list)->list:
    n = len(arr)
    n_cnt = 0
    for i in range(n):
        if arr[i] != 0:
            n_cnt += 1
            arr[n_cnt - 1] = arr[i]
    for j in range(n_cnt,n):
        arr[j] = 0
    return arr
#takes O(n) time only O(1) space
```

```python
move_zeros_fast([1,0,5,6,0,0,3,4,0])
```

```python
move_zeros_fast([0,0,0,0,1,0,5,6,0,0,3,4,0])
```

```python
move_zeros_fast([1,2,3,4,0,0,0,0,1,0,5,6,0,0,3,4,0])
```

```python
def move_zeros_fast_another(arr:list)->list:
    n = len(arr)
    n_cnt = 0
    for i in range(n):
        if arr[i] != 0:
            arr[i], arr[n_cnt] = arr[n_cnt], arr[i]
            n_cnt += 1
    return arr
#takes O(n) time only O(1) space
```

```python
move_zeros_fast_another([1,2,3,4,0,0,0,0,1,0,5,6,0,0,3,4,0])
```
